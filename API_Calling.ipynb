```python
#api calling is application programming interface it means it work like messenger or bridge between your application and other service
#suppose yp
import json
import requests
```

```python
#api to fetch tempreture of a city

city_name=input('enter a city name:')
api_key="6f009dadba5a623299c5efe5b0c3ffa8"


#to built api
api_url= f'https://api.openweathermap.org/data/2.5/weather?q={city_name}&appid={api_key}&units=metric'
get_server_information=requests.get(api_url)
json_data=get_server_information.json()
```

```python
#prity print json string
#means how can you make json format data into most human friendly is the prety print
prety_data=json.dumps(json_data,indent=4)
```
